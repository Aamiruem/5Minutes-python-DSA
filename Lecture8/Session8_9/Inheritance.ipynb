```python
# Parent class
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def make_sound(self):
        return "Some generic animal sound"

    def describe(self):
        return f"{self.name} is a {self.species}"


# Child class
class Dog(Animal):
    def __init__(self, name, breed):
        # Call parent's __init__
        super().__init__(name, species="Canine")
        self.breed = breed

    # Method overriding
    def make_sound(self):
        return "Woof! Woof!"

    def fetch(self):
        return f"{self.name} is fetching the ball"


# Usage
dog = Dog("Buddy", "Golden Retriever")
print(dog.describe())  # Buddy is a Canine
print(dog.make_sound())  # Woof! Woof!
print(dog.fetch())  # Buddy is fetching the ball
```

```python
class Father:
    def __init__(self):
        self.father_name = "John"

    def skills(self):
        return "Programming, Mathematics"

    def height(self):
        return "6 feet"


class Mother:
    def __init__(self):
        self.mother_name = "Mary"

    def skills(self):
        return "Cooking, Painting"

    def height(self):
        return "5 feet 4 inches"


class Child(Father, Mother):
    def __init__(self, name):
        Father.__init__(self)
        Mother.__init__(self)
        self.name = name

    def skills(self):
        # Access both parents' skills
        father_skills = Father.skills(self)
        mother_skills = Mother.skills(self)
        return f"Father: {father_skills}, Mother: {mother_skills}"


# MRO demonstration
child = Child("Alex")
print(child.name)  # Alex
print(child.skills())  # Father: Programming, Mathematics, Mother: Cooking, Painting
print(child.height())  # 6 feet (takes from Father due to MRO)
print(Child.__mro__)  # Shows Method Resolution Order
```

# Multilevel Inheritance 
 
### Inheritance = Child class gets everything from Parent class, like a child inherits property from parents.

```python
class LivingBeing:
    def breathe(self):
        return "Breathing Oxygen"

class Animal(LivingBeing):
    def move(self):
        return "Moving Around"
    
class Bird(Animal):
    def fly(self):
        return "Flying in the sky"
    
    
class Sparrow(Bird):
    def chirp(self):
        return "Sound Like chirp! chirp!" # chirp means chahchahat
    
    
sparrow = Sparrow()
print(sparrow.breathe())
print(sparrow.move())
print(sparrow.fly())
print(sparrow.chirp())
```

```python
from datetime import datetime

class Employee:
    raise_amount = 1.04  # Class variable
    
    def __init__(self, name, salary, join_date):
        self.name = name
        self.salary = salary
        self.join_date = join_date
    
    def apply_raise(self):
        self.salary = int(self.salary * self.raise_amount)
        return self.salary
    
    def get_tenure(self):
        tenure = datetime.now() - self.join_date
        return tenure.days // 365
    
    def __str__(self):
        return f"{self.name} - {self.designation}"

class Manager(Employee):
    raise_amount = 1.10  # Override class variable
    
    def __init__(self, name, salary, join_date, team_size):
        super().__init__(name, salary, join_date)
        self.team_size = team_size
        self.designation = "Manager"
    
    def manage_team(self):
        return f"Managing {self.team_size} employees"

class Developer(Employee):
    raise_amount = 1.08
    
    def __init__(self, name, salary, join_date, programming_languages):
        super().__init__(name, salary, join_date)
        self.programming_languages = programming_languages
        self.designation = "Developer"
    
    def add_language(self, language):
        if language not in self.programming_languages:
            self.programming_languages.append(language)
    
    def get_languages(self):
        return ", ".join(self.programming_languages)

class SeniorDeveloper(Developer):
    raise_amount = 1.12
    
    def __init__(self, name, salary, join_date, programming_languages, mentees):
        super().__init__(name, salary, join_date, programming_languages)
        self.mentees = mentees
        self.designation = "Senior Developer"
    
    def mentor(self):
        return f"Mentoring {len(self.mentees)} junior developers"

# Usage
manager = Manager("John Doe", 80000, datetime(2020, 1, 15), 10)
dev = Developer("Jane Smith", 65000, datetime(2021, 6, 1), ["Python", "JavaScript"])
senior_dev = SeniorDeveloper("Bob Wilson", 90000, datetime(2019, 3, 10), 
                             ["Python", "Java", "Go"], ["Alice", "Charlie"])

print(manager.apply_raise())        # 88000
print(dev.apply_raise())            # 70200
print(senior_dev.apply_raise())     # 100800

print(isinstance(senior_dev, Employee))    # True
print(isinstance(senior_dev, Developer))   # True
print(isinstance(manager, Developer))      # False
print(issubclass(SeniorDeveloper, Employee))  # True
```

```python

class Shape:
    def __init__(self, color):
        self.color = color
        print(f"shape initialized with color {color}")
        
    def area(self):
        return 0

class Rectangle(Shape):
    def __init__(self, color, width, height):
        super().__init__(color)
        self.width = width
        self.height  = height
        print(f"Rectangle created:  {width} * {height}")
        
    def area(self):
        return self.width * self.height
    
class Square(Rectangle):
    def __init__(self, color, side):
        super().__init__(color, side, side)
        print(f"square created with side: {side}")
        
        
# demonstrate super() chain
square = Square("Red", 5)
print(f"Area: {square.area()}")
print(f"color: {square.color}")
    
```

```python
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    def __init__(self, amount):
        self.amount = amount
    
    @abstractmethod
    def process_payment(self):
        """Every payment processor must implement this method"""
        pass
    
    @abstractmethod
    def refund(self):
        """Every payment processor must implement this method"""
        pass
    
    # Concrete method (has implementation)
    def validate_amount(self):
        return self.amount > 0

class CreditCardPayment(PaymentProcessor):
    def __init__(self, amount, card_number):
        super().__init__(amount)
        self.card_number = card_number
    
    def process_payment(self):
        return f"Processing ${self.amount} from card {self.card_number[-4:]}"
    
    def refund(self):
        return f"Refunding ${self.amount} to card {self.card_number[-4:]}"

class PayPalPayment(PaymentProcessor):
    def __init__(self, amount, email):
        super().__init__(amount)
        self.email = email
    
    def process_payment(self):
        return f"Processing ${self.amount} via PayPal from {self.email}"
    
    def refund(self):
        return f"Refunding ${self.amount} to {self.email}"

# This would raise TypeError - can't instantiate abstract class
# payment = PaymentProcessor(100)

# Correct usage
credit = CreditCardPayment(150, "1234-5678-9012-3456")
paypal = PayPalPayment(75, "user@example.com")

print(credit.process_payment())  # Processing $150 from card 3456
print(paypal.process_payment())  # Processing $75 via PayPal from user@example.com
```

```python
class A:
    def method(self):
        return "A Method "
    
    def common(self):
        return "common from A"

class B(A):
    def method(self):
        return "B Method "
    
    def common (self):
        return  " common from B "
    
    
class C(A):
    def method(self):
        return " C Method "
    
    def common (self):
        return " common from C "
    
class D(B, C):
        def common (self):
            b_result = B.common(self)
            
            c_result = C.common(self)
            
            return f"{b_result} and {c_result}"
        
d = D()
print(D.__mro__)
print(d.method)
print(d.common)
```

```python
class A:
    def method(self):
        return "A method"
    
    def common(self):
        return "Common from A"

class B(A):
    def method(self):
        return "B method"
    
    def common(self):
        return "Common from B"

class C(A):
    def method(self):
        return "C method"
    
    def common(self):
        return "Common from C"

class D(B, C):
    def common(self):
        # Calling specific parent's method
        b_result = B.common(self)
        c_result = C.common(self)
        return f"{b_result} and {c_result}"

# MRO in diamond problem
d = D()
print(D.__mro__)  # D -> B -> C -> A -> object
print(d.method())  # B method (from B due to MRO)
print(d.common())  # Common from B and Common from C
```
