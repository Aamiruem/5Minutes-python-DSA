```python
class Father:
    def __init__(self):
        self.father_name = "John"

    def skills(self):
        return "Programming, Mathematics"

    def height(self):
        return "6 feet"


class Mother:
    def __init__(self):
        self.mother_name = "Mary"

    def skills(self):
        return "Cooking, Painting"

    def height(self):
        return "5 feet 4 inches"


class Child(Father, Mother):
    def __init__(self, name):
        Father.__init__(self)
        Mother.__init__(self)
        self.name = name

    def skills(self):
        # Access both parents' skills
        father_skills = Father.skills(self)
        mother_skills = Mother.skills(self)
        return f"Father: {father_skills}, Mother: {mother_skills}"


# MRO demonstration
child = Child("Alex")
print(child.name)  # Alex
print(child.skills())  # Father: Programming, Mathematics, Mother: Cooking, Painting
print(child.height())  # 6 feet (takes from Father due to MRO)
print(Child.__mro__)  # Shows Method Resolution Order
```

```python
class A:
    def method(self):
        return "A Method "
    
    def common(self):
        return "common from A"

class B(A):
    def method(self):
        return "B Method "
    
    def common (self):
        return  " common from B "
    
    
class C(A):
    def method(self):
        return " C Method "
    
    def common (self):
        return " common from C "
    
class D(B, C):
        def common (self):
            b_result = B.common(self)
            
            c_result = C.common(self)
            
            return f"{b_result} and {c_result}"
        
d = D()
print(D.__mro__)
print(d.method)
print(d.common)
```

```python
class A:
    def method(self):
        return "A method"
    
    def common(self):
        return "Common from A"

class B(A):
    def method(self):
        return "B method"
    
    def common(self):
        return "Common from B"

class C(A):
    def method(self):
        return "C method"
    
    def common(self):
        return "Common from C"

class D(B, C):
    def common(self):
        # Calling specific parent's method
        b_result = B.common(self)
        c_result = C.common(self)
        return f"{b_result} and {c_result}"

# MRO in diamond problem
d = D()
print(D.__mro__)  # D -> B -> C -> A -> object
print(d.method())  # B method (from B due to MRO)
print(d.common())  # Common from B and Common from C
```

```python
class BankAccount:
    def __init__(self, balance):
        # private attribute
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal amount.")

    def get_balance(self):
        return self.__balance


# Testing the class
acc = BankAccount(1000)
print("Initial balance:", acc.get_balance())

acc.deposit(500)
print("After deposit:", acc.get_balance())

acc.withdraw(200)
print("After withdrawal:", acc.get_balance())

acc.withdraw(2000)  # invalid withdrawal
print("After invalid withdrawal:", acc.get_balance())

acc.deposit(-100)   # invalid deposit
print("After invalid deposit:", acc.get_balance())
```

```python
# WITHOUT Encapsulation (Bad)
class BankAccount:
    def __init__(self, balance):
        self.balance = balance  # Directly accessible - DANGEROUS!

# Anyone can change balance directly
acc = BankAccount(1000)
acc.balance = -500  # BAD! Can set negative balance

# WITH Encapsulation (Good)
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private (hidden)
    
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
    
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
    
    def get_balance(self):
        return self.__balance

# Dry Run:
acc = BankAccount(1000)
print(acc.get_balance())  # 1000
acc.deposit(500)          # balance becomes 1500
acc.withdraw(200)         # balance becomes 1300
# acc.__balance - ERROR! Can't access directly
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    
    @abstractmethod
    def perimeter(self):
        pass
    
    def display(self):
        print(" This is a Shape ")
        
        
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
        
    def area(self):
        return self.length*self.width
    
    def perimeter(self):
        return 2* (self.length + self.width)
    
class circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        
        
    def area(self):
        return 3.14 * self.radius * self.radius
    
    
    def perimeter(self):
        return 2* 3.14 * self.radius
    
    
    
react = Rectangle(5, 3)
print(react.area())
print(react.perimeter())
    
react.display()
        
```

```python
from abc import ABC, abstractmethod
from typing import List
from datetime import datetime
import uuid

class User:
    def __init__(self, name: str, email: str):
        self.id = str(uuid.uuid4())
        self.name = name
        self.email = email
        self.cart = ShoppingCart()
        self.orders = []
    
    def add_to_cart(self, product, quantity=1):
        self.cart.add_item(product, quantity)
    
    def checkout(self):
        if self.cart.items:
            order = Order(self, self.cart)
            self.orders.append(order)
            self.cart.clear()
            return order
        return None

class Product:
    def __init__(self, name: str, price: float, stock: int):
        self.id = str(uuid.uuid4())
        self.name = name
        self.price = price
        self.stock = stock
    
    def reduce_stock(self, quantity: int):
        if quantity > self.stock:
            raise ValueError(f"Insufficient stock for {self.name}")
        self.stock -= quantity

class ShoppingCart:
    def __init__(self):
        self.items = {}  # product_id -> (product, quantity)
    
    def add_item(self, product: Product, quantity: int):
        if product.id in self.items:
            current_qty = self.items[product.id][1]
            self.items[product.id] = (product, current_qty + quantity)
        else:
            self.items[product.id] = (product, quantity)
    
    def remove_item(self, product_id: str):
        if product_id in self.items:
            del self.items[product_id]
    
    def get_total(self):
        return sum(p.price * qty for p, qty in self.items.values())
    
    def clear(self):
        self.items.clear()

class Payment(ABC):
    @abstractmethod
    def process(self, amount: float) -> bool:
        pass

class CreditCardPayment(Payment):
    def __init__(self, card_number: str, expiry: str, cvv: str):
        self.card_number = card_number
        self.expiry = expiry
        self.cvv = cvv
    
    def process(self, amount: float) -> bool:
        # Simulate payment processing
        print(f"Processing ${amount} via Credit Card ending in {self.card_number[-4:]}")
        return True

class PayPalPayment(Payment):
    def __init__(self, email: str):
        self.email = email
    
    def process(self, amount: float) -> bool:
        print(f"Processing ${amount} via PayPal account {self.email}")
        return True

class Order:
    def __init__(self, user: User, cart: ShoppingCart):
        self.id = str(uuid.uuid4())
        self.user = user
        self.items = dict(cart.items)
        self.total = cart.get_total()
        self.date = datetime.now()
        self.status = "pending"
    
    def pay(self, payment: Payment):
        if payment.process(self.total):
            self.status = "paid"
            for product, qty in self.items.values():
                product.reduce_stock(qty)
            return True
        return False

# Usage
# Create users
alice = User("Alice", "alice@email.com")
bob = User("Bob", "bob@email.com")

# Add products
laptop = Product("Laptop", 999.99, 10)
phone = Product("Phone", 599.99, 20)
book = Product("Book", 29.99, 50)

# Alice shops
alice.add_to_cart(laptop, 1)
alice.add_to_cart(book, 2)

# Checkout
order = alice.checkout()
if order:
    payment = CreditCardPayment("1234567890123456", "12/25", "123")
    if order.pay(payment):
        print(f"Order {order.id} paid successfully!")
        print(f"Total: ${order.total}")
        print(f"Items: {len(order.items)}")

# Bob shops differently
bob.add_to_cart(phone, 1)
order_bob = bob.checkout()
if order_bob:
    payment_bob = PayPalPayment("bob@email.com")
    order_bob.pay(payment_bob)
    print(f"Bob's order total: ${order_bob.total}")
```

```python
from abc import ABC, abstractmethod
import random

class Character(ABC):
    def __init__(self, name, health=100, attack=10, defense=5):
        self.name = name
        self.max_health = health
        self.health = health
        self.attack = attack
        self.defense = defense
        self.level = 1
        self.experience = 0
        self.inventory = []
    
    @abstractmethod
    def special_ability(self):
        pass
    
    def take_damage(self, damage):
        actual_damage = max(0, damage - self.defense)
        self.health -= actual_damage
        return actual_damage
    
    def is_alive(self):
        return self.health > 0
    
    def attack_enemy(self, enemy):
        damage = self.attack + random.randint(-2, 5)
        actual_damage = enemy.take_damage(damage)
        return f"{self.name} deals {actual_damage} damage to {enemy.name}"
    
    def gain_experience(self, exp):
        self.experience += exp
        if self.experience >= self.level * 100:
            self.level_up()
    
    def level_up(self):
        self.level += 1
        self.max_health += 20
        self.health = self.max_health
        self.attack += 5
        self.defense += 3
        print(f"{self.name} leveled up to {self.level}!")

class Warrior(Character):
    def __init__(self, name):
        super().__init__(name, health=150, attack=15, defense=10)
    
    def special_ability(self):
        self.attack *= 2
        return f"{self.name} uses Rage! Attack doubled for this turn!"

class Mage(Character):
    def __init__(self, name):
        super().__init__(name, health=80, attack=25, defense=3)
        self.mana = 100
    
    def special_ability(self):
        if self.mana >= 50:
            self.mana -= 50
            return f"{self.name} casts Fireball! Massive damage!"
        return "Not enough mana"

class Archer(Character):
    def __init__(self, name):
        super().__init__(name, health=100, attack=20, defense=5)
        self.arrows = 20
    
    def special_ability(self):
        if self.arrows >= 5:
            self.arrows -= 5
            self.attack += 10
            return f"{self.name} uses Multi-shot! Attack increased!"
        return "Not enough arrows"

class Enemy:
    def __init__(self, name, health, attack, defense, exp_reward):
        self.name = name
        self.health = health
        self.attack = attack
        self.defense = defense
        self.exp_reward = exp_reward
    
    def take_damage(self, damage):
        actual_damage = max(0, damage - self.defense)
        self.health -= actual_damage
        return actual_damage
    
    def is_alive(self):
        return self.health > 0

class Game:
    def __init__(self):
        self.player = None
        self.enemies = []
    
    def create_player(self):
        print("Choose your class:")
        print("1. Warrior (Tanky)")
        print("2. Mage (High damage)")
        print("3. Archer (Balanced)")
        
        choice = input("Enter choice (1-3): ")
        name = input("Enter character name: ")
        
        if choice == "1":
            self.player = Warrior(name)
        elif choice == "2":
            self.player = Mage(name)
        elif choice == "3":
            self.player = Archer(name)
        else:
            print("Invalid choice, defaulting to Warrior")
            self.player = Warrior(name)
    
    def spawn_enemy(self):
        enemy_types = [
            ("Goblin", 50, 8, 2, 50),
            ("Orc", 80, 12, 5, 80),
            ("Troll", 120, 10, 8, 120),
            ("Dragon", 200, 25, 10, 300)
        ]
        enemy_data = random.choice(enemy_types)
        enemy = Enemy(*enemy_data)
        print(f"A wild {enemy.name} appears!")
        return enemy
    
    def battle(self, enemy):
        print(f"\n=== Battle with {enemy.name} ===")
        battle_log = []
        
        while self.player.is_alive() and enemy.is_alive():
            print(f"\n{self.player.name}: {self.player.health}/{self.player.max_health} HP")
            print(f"{enemy.name}: {enemy.health} HP")
            
            print("\nActions:")
            print("1. Attack")
            print("2. Special Ability")
            print("3. Defend (reduce damage)")
            
            choice = input("Choose action: ")
            
            if choice == "1":
                result = self.player.attack_enemy(enemy)
                battle_log.append(result)
                print(result)
            
            elif choice == "2":
                result = self.player.special_ability()
                battle_log.append(result)
                print(result)
                if "Fireball" in result or "Rage" in result or "Multi-shot" in result:
                    result = self.player.attack_enemy(enemy)
                    battle_log.append(result)
                    print(result)
            
            elif choice == "3":
                self.player.defense += 5
                print(f"{self.player.name} defends!")
            
            if enemy.is_alive():
                enemy_damage = enemy.attack + random.randint(-3, 3)
                actual_damage = self.player.take_damage(enemy_damage)
                print(f"{enemy.name} deals {actual_damage} damage")
        
        if self.player.is_alive():
            print(f"\nVictory! {self.player.name} defeated {enemy.name}!")
            self.player.gain_experience(enemy.exp_reward)
            print(f"Gained {enemy.exp_reward} experience!")
            return True
        else:
            print(f"\nGame Over! {self.player.name} was defeated...")
            return False
    
    def play(self):
        print("=== Welcome to RPG Adventure ===")
        self.create_player()
        print(f"\nWelcome, {self.player.name}!")
        
        round_num = 1
        while self.player.is_alive():
            print(f"\n--- Round {round_num} ---")
            enemy = self.spawn_enemy()
            if not self.battle(enemy):
                break
            round_num += 1
            
            if input("\nContinue adventuring? (y/n): ").lower() != 'y':
                break
        
        print(f"\nGame Over! {self.player.name} reached Level {self.player.level}")

# Start game
# game = Game()
# game.play()
```

```python
class Point:
    def __init__(self, x, y):
        self.x  = x
        self.y = y
        
    def __add__(self, other):
        return Point(self.x +  other.x, self.y + other.y)
    
    def __str__(self):
        return f"({self.x}, {self.y})"
    
p1 = Point(2, 3)
p2 = Point(4, 5)
p3 = p1 + p2

print(p3)
```

```python
class Book:
    def __init__(self, tittle, pages):
        self.tittle = tittle
        self.pages = pages

    def __len__(self):
        return self.pages

    def __str__(self):
        return f"Book: {self.tittle}"

    def __eq__(self, other):
        return self.pages == other.pages


book1 = Book("Python", 300)
book2 = Book("Java", 400)
print(len(book1))
print(book1)
print(book1 == book2)
print(book1 == book2)
        
        
```
