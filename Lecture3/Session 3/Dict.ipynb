🟢 4. Dictionary vs List

This is very important.

List


A list normally uses a position/index.

students = ["Aamir", "Rahul", "John"]


print(students[0])

Output:

Aamir

You say:

Give me item at position 0
Dictionary

A dictionary uses a key.

student = {
    "name": "Aamir",
    "age": 22
}


print(student["name"])

Output:

Aamir

You say:

Give me the value associated with "name"
Visual comparison
LIST


Index       Value
  0    →    Aamir
  1    →    Rahul
  2    →    John






DICTIONARY


Key         Value
"name"  →   Aamir
"age"   →   22

🧠 Easy memory trick:

List → position

Dictionary → name/key

```python
name = {"n": ["aamir", "john"]}

name
```

### 🟢 12. pop()

pop() removes a key and returns its value.

```python
student = {"name": "aamir", "age": 24}

age = student.pop("age")

print(age)

print(student)
```

### 🟢 13. popitem()

Removes and returns the last inserted key-value pair.

```python
student = {"name": "Aamir", "age": 22, "course": "Python"}

item = student.popitem()
print(item)
print(student)
```

### 🟢 14. clear()

Removes everything.

```python
student = {"name": "Aamir", "age": 22}

student.clear()

print(student)
```

### 🟢 15. keys()

Returns dictionary keys.

```python
student = {"name": "Aamir", "age": 22}

print(student.keys())
```

### 🟢 16. values()

Returns all values.

```python
student = {"name": "Aamir", "age": 22}

for value1 in student.values():
    print(value1)
```

### 🟢 17. items()

This is extremely important for loops.

```python
student = {"name": "Aamir", "age": 22}

for key, value in student.items():
    print(key, value)
```

### 🟢 18. Looping Through a Dictionary
Loop through keys

```python
student = {"name": "Aamir", "age": 22}


for key in student:
    print(key)
```

```python
for value in student.values():
    print(value)
```

```python
for key, value in student.items():
    print(key, " = ", value)
```

### 🟨 19. The in Operator

You can check whether a key exists.

```python
student = {"name": "Aamir", "age": 22}

print("name" in student)
```

```python
print("Aamir" in student)
```

But:

print("Aamir" in student)

is:

False

Why?

in checks keys, not values.

"name"  → Aamir
   ↑
 checks key


"Aamir"
   ↑
 value, not key

To search values:

print("Aamir" in student.values())

Output:

True

### 🟢 28. copy()

To create a shallow copy:

```python
a = {"name": "Aamir"}

b = a.copy()

b["name"] = "John"

print(a)
print(b)
```

### 🟢 30. fromkeys()

Creates a dictionary from keys with the same initial value.

```python
keys = ["name", "age", "city"]

data = dict.fromkeys(keys, "unknown")

print(data)
```
