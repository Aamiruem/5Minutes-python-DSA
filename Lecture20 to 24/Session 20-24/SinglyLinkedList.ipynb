```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
class sinlyLinkedList:
    def __init__(self):
        self.head = None
        self.size = 0


# check if list is empty
def is_empty(self):
    return self.head is None


# Get size of List
def get_size(self):
    return self.size


# insert at begnning -o(1)


def insert_at_begnning(self, data):
    new_node = Node(data)
    if self.is_empty():
        self.head = new_node
    else:
        current = self.head
        while current.next:
            current = current.next
            current.next = new_node
            self.size += 1
```

# insert at end o(n)

```python
def insert_at_end(self, data):
    new_node = Node(data)
    if self.is_empty():
        self.head = new_node
    else:
        current = self.head
        while current.next:
            current = current.next
            current.next = new_node
    self.size += 1
```
