# Q1: Create & Display Linked List
Problem: Build a linked list from array and display it.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def create_from_array(self, arr):
        if not arr:
            return None

        self.head = Node(arr[0])
        current = self.head

        for i in range(1, len(arr)):
            current.next = Node(arr[i])
            current = current.next
        return self.head

    def display(self):
        elements = []
        current = self.head
        while current:
            elements.append(current.data)
            current = current.next
        return elements

    def insert_at_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def delete_from_end(self):
        if not self.head:
            return None

        if not self.head.next:
            deleted_data = self.head.data
            self.head = None
            return deleted_data

        current = self.head
        while current.next.next:
            current = current.next
            deleted_data = current.next.data
            current.next = None

            return deleted_data


# crete linked list from array
ll = LinkedList()
ll.create_from_array([10, 20, 30, 40])
print("Original List: ", ll.display())

# Insert at end
ll.insert_at_end(50)
print("After inserting 50: ", ll.display())

# Delete from end
deleted = ll.delete_from_end()
print("delete element:", deleted)
print("List after deletion:", ll.display())

print("just deletion ", ll.display())


# delete until empty
ll.delete_from_end()
ll.delete_from_end()
ll.delete_from_end()
ll.delete_from_end()
print("List after deleting all:", ll.display())
```

# Q2: Count Nodes

```python
def count_nodes(head):
    count = 0
    current = head
    while current:
        count += 1
        current = current.next
    return count


def count_node_recuresive(head):
    if not head:
        return 0
    return 1 + count_node_recuresive(head.next)
```

```python
# Method 1: Iterative (3 Pointers)
def reverse_iterative(head):
    """Reverse using 3 pointers"""
    prev = None
    current = head

    while current:
        next_node = current.next  # Save next
        current.next = prev  # Reverse link
        prev = current  # Move prev
        current = next_node  # Move current

    return prev


# Dry Run: [1] → [2] → [3] → None
"""
Initial: prev=None, current=1
┌─────┐    ┌─────┐    ┌─────┐
│  1  │───→│  2  │───→│  3  │───→ None
└─────┘    └─────┘    └─────┘
  ↑
current

Iteration 1:
next_node = 2 (save)
1.next = None (reverse)
prev = 1, current = 2

None ← [1]    [2] → [3] → None
        ↑      ↑
       prev  current

Iteration 2:
next_node = 3
2.next = 1
prev = 2, current = 3

None ← [1] ← [2]    [3] → None
              ↑      ↑
             prev  current

Iteration 3:
next_node = None
3.next = 2
prev = 3, current = None

None ← [1] ← [2] ← [3]
                    ↑
                   prev

Return prev (3)

Result: [3] → [2] → [1] → None
"""


# Method 2: Recursive
def reverse_recursive(head):
    """Reverse recursively"""
    # Base case
    if not head or not head.next:
        return head

    # Reverse rest of list
    new_head = reverse_recursive(head.next)

    # Reverse current node
    head.next.next = head
    head.next = None

    return new_head


# Dry Run: [1] → [2] → [3]
"""
Call 1: reverse(1)
  Call 2: reverse(2)
    Call 3: reverse(3) → returns 3 (base case)
  
  Unwind Call 2: 
  2.next.next = 2 → 3.next = 2
  2.next = None
  Returns 3

Unwind Call 1:
1.next.next = 1 → 2.next = 1
1.next = None
Returns 3

Result: 3 → 2 → 1 → None
"""


# Method 3: Stack
def reverse_stack(head):
    """Reverse using stack"""
    if not head:
        return None

    stack = []
    current = head
    while current:
        stack.append(current)
        current = current.next

    head = stack.pop()
    current = head
    while stack:
        current.next = stack.pop()
        current = current.next
    current.next = None

    return head


# Dry Run: [1] → [2] → [3]
"""
Push all to stack:
stack = [1, 2, 3] (top is 3)

Pop 3 → head = 3
Pop 2 → 3.next = 2
Pop 1 → 2.next = 1
Set 1.next = None

Result: [3] → [2] → [1] → None
"""
```
