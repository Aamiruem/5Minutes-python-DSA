```python
# Method 1: Iterative (3 Pointers)
def reverse_iterative(head):
    """Reverse using 3 pointers"""
    prev = None
    current = head

    while current:
        next_node = current.next  # Save next
        current.next = prev  # Reverse link
        prev = current  # Move prev
        current = next_node  # Move current

    return prev


# Dry Run: [1] → [2] → [3] → None
"""
Initial: prev=None, current=1
┌─────┐    ┌─────┐    ┌─────┐
│  1  │───→│  2  │───→│  3  │───→ None
└─────┘    └─────┘    └─────┘
  ↑
current

Iteration 1:
next_node = 2 (save)
1.next = None (reverse)
prev = 1, current = 2

None ← [1]    [2] → [3] → None
        ↑      ↑
       prev  current

Iteration 2:
next_node = 3
2.next = 1
prev = 2, current = 3

None ← [1] ← [2]    [3] → None
              ↑      ↑
             prev  current

Iteration 3:
next_node = None
3.next = 2
prev = 3, current = None

None ← [1] ← [2] ← [3]
                    ↑
                   prev

Return prev (3)

Result: [3] → [2] → [1] → None
"""


# Method 2: Recursive
def reverse_recursive(head):
    """Reverse recursively"""
    # Base case
    if not head or not head.next:
        return head

    # Reverse rest of list
    new_head = reverse_recursive(head.next)

    # Reverse current node
    head.next.next = head
    head.next = None

    return new_head


# Dry Run: [1] → [2] → [3]
"""
Call 1: reverse(1)
  Call 2: reverse(2)
    Call 3: reverse(3) → returns 3 (base case)
  
  Unwind Call 2: 
  2.next.next = 2 → 3.next = 2
  2.next = None
  Returns 3

Unwind Call 1:
1.next.next = 1 → 2.next = 1
1.next = None
Returns 3

Result: 3 → 2 → 1 → None
"""


# Method 3: Stack
def reverse_stack(head):
    """Reverse using stack"""
    if not head:
        return None

    stack = []
    current = head
    while current:
        stack.append(current)
        current = current.next

    head = stack.pop()
    current = head
    while stack:
        current.next = stack.pop()
        current = current.next
    current.next = None

    return head


# Dry Run: [1] → [2] → [3]
"""
Push all to stack:
stack = [1, 2, 3] (top is 3)

Pop 3 → head = 3
Pop 2 → 3.next = 2
Pop 1 → 2.next = 1
Set 1.next = None

Result: [3] → [2] → [1] → None
"""
```

## Q8: Find Nth Node from End

```python
def nth_from_end(head, n):
    """Find nth node from end using gap method"""
    if not head or n <= 0:
        return None

    first = head
    second = head

    # Create gap of n nodes
    for _ in range(n):
        if not first:
            return None  # n is larger than list
        first = first.next

    # Move both until first reaches end
    while first:
        first = first.next
        second = second.next

    return second.data


# Dry Run: [1] → [2] → [3] → [4] → [5], n=2 (find 2nd from end)
"""
Step 1: Create gap
first moves 2 steps: first at 3
second at 1

[1] → [2] → [3] → [4] → [5]
 ↑            ↑
second      first

Step 2: Move both
first=4, second=2
first=5, second=3
first=None, second=4, stop

Return 4 (2nd from end)
"""
```

## Q9: Detect and Remove Cycle

```python
def detect_cycle(head):
    """Floyd's Cycle Detection"""
    if not head or not head.next:
        return None

    slow = head
    fast = head

    # Detect cycle
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return slow  # Meeting point

    return None


def remove_cycle(head):
    """Detect and remove cycle"""
    # Detect cycle
    meeting_point = detect_cycle(head)
    if not meeting_point:
        return False

    # Find cycle start
    slow = head
    while slow != meeting_point:
        slow = slow.next
        meeting_point = meeting_point.next

    cycle_start = slow

    # Find last node in cycle
    current = cycle_start
    while current.next != cycle_start:
        current = current.next

    # Break cycle
    current.next = None
    return True


# Dry Run: [1] → [2] → [3] → [4] → [5] → [3] (cycle)
"""
Cycle Detection:
Step 1: slow=2, fast=3
Step 2: slow=3, fast=5
Step 3: slow=4, fast=4 (meet!)

Cycle Start:
slow=head(1), meeting=4
Step 1: slow=2, meeting=5
Step 2: slow=3, meeting=3 (meet at cycle start)

Remove:
Find last node (5)
5.next = None

Result: [1] → [2] → [3] → [4] → [5] → None
"""
```

## Q10: Merge Two Sorted Lists

```python
def merge_sorted(l1, l2):
    """Merge two sorted linked lists"""
    dummy = Node(0)
    current = dummy

    while l1 and l2:
        if l1.data <= l2.data:
            current.next = l1
            l1 = l1.next
        else:
            current.next = l2
            l2 = l2.next
        current = current.next

    # Attach remaining
    if l1:
        current.next = l1
    if l2:
        current.next = l2

    return dummy.next


# Dry Run: l1=[1,3,5], l2=[2,4,6]
"""
dummy → None
current = dummy

Step 1: 1≤2 → current.next=1, l1=3
dummy → 1

Step 2: 3≤2? No → current.next=2, l2=4
dummy → 1 → 2

Step 3: 3≤4 → current.next=3, l1=5
dummy → 1 → 2 → 3

Step 4: 5≤4? No → current.next=4, l2=6
dummy → 1 → 2 → 3 → 4

Step 5: 5≤6 → current.next=5, l1=None
dummy → 1 → 2 → 3 → 4 → 5

l1 is None, attach l2
dummy → 1 → 2 → 3 → 4 → 5 → 6

Result: [1,2,3,4,5,6]
"""
```

## Q11: Palindrome Linked List

```python
def is_palindrome(head):
    """Check if linked list is palindrome"""
    if not head or not head.next:
        return True

    # Find middle
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    # Reverse second half
    prev = None
    while slow:
        next_node = slow.next
        slow.next = prev
        prev = slow
        slow = next_node

    # Compare first and second half
    first = head
    second = prev
    while second:
        if first.data != second.data:
            return False
        first = first.next
        second = second.next

    return True


# Dry Run: [1,2,3,2,1]
"""
Step 1: Find middle
slow=1, fast=1
slow=2, fast=3
slow=3, fast=5 (fast.next is None, stop)
middle = 3

Step 2: Reverse second half [3,2,1] → [1,2,3]
Original: 1 → 2 → 3 → 2 → 1
After reverse: 1 → 2 → 3 → 1 → 2

Step 3: Compare
1==1 ✓
2==2 ✓
3==3 ✓ (stop at None)

Return True
"""
```

## Q12: Intersection of Two Linked Lists

```python
def get_intersection(headA, headB):
    """Find intersection point of two linked lists"""
    if not headA or not headB:
        return None

    # Get lengths
    lenA, lenB = 0, 0
    currA, currB = headA, headB

    while currA:
        lenA += 1
        currA = currA.next

    while currB:
        lenB += 1
        currB = currB.next

    # Reset pointers
    currA, currB = headA, headB

    # Move longer list pointer
    if lenA > lenB:
        for _ in range(lenA - lenB):
            currA = currA.next
    else:
        for _ in range(lenB - lenA):
            currB = currB.next

    # Find intersection
    while currA and currB:
        if currA == currB:
            return currA
        currA = currA.next
        currB = currB.next

    return None


# Dry Run:
"""
List A: 1 → 2 → 3 → 4 → 5
List B: 6 → 4 → 5 (4 is intersection)

lenA = 5, lenB = 3
Move A by 2: A at 3, B at 6

Compare:
3≠6, move both
4=4 ✓ (intersection found)
"""
```

## Q13: Rotate Linked List

```python
def rotate_right(head, k):
    """Rotate list right by k positions"""
    if not head or not head.next or k == 0:
        return head

    # Find length and last node
    length = 1
    last = head
    while last.next:
        last = last.next
        length += 1

    # Adjust k
    k = k % length
    if k == 0:
        return head

    # Find new last (length - k - 1 position)
    current = head
    for _ in range(length - k - 1):
        current = current.next

    # Rotate
    new_head = current.next
    current.next = None
    last.next = head

    return new_head


# Dry Run: [1,2,3,4,5], k=2
"""
length=5, k=2%5=2
Find new last: position 5-2-1=2 (node 3)

[1] → [2] → [3] → [4] → [5]
             ↑      ↑       ↑
          current  new     last

Step 1: new_head = 3.next = 4
Step 2: 3.next = None
Step 3: 5.next = 1 (old head)

Result: [4] → [5] → [1] → [2] → [3]
"""
```

## Q14: Remove Duplicates

```python
def remove_duplicates_sorted(head):
    """Remove duplicates from sorted list"""
    current = head

    while current and current.next:
        if current.data == current.next.data:
            current.next = current.next.next
        else:
            current = current.next

    return head


def remove_duplicates_unsorted(head):
    """Remove duplicates from unsorted list using hash set"""
    if not head:
        return None

    seen = set()
    current = head
    seen.add(current.data)

    while current.next:
        if current.next.data in seen:
            current.next = current.next.next
        else:
            seen.add(current.next.data)
            current = current.next

    return head


# Dry Run: Remove duplicates from [1,1,2,3,3]
"""
Sorted:
Step 1: current=1, next=1, match → 1.next=2
[1] → [2] → [3] → [3]

Step 2: current=1, next=2, no match → move
Step 3: current=2, next=3, no match → move
Step 4: current=3, next=3, match → 3.next=None

Result: [1,2,3]
"""
```

## Q15: Swap Nodes in Pairs

```python
def swap_pairs(head):
    """Swap every two adjacent nodes"""
    dummy = Node(0)
    dummy.next = head
    prev = dummy

    while prev.next and prev.next.next:
        # Nodes to swap
        first = prev.next
        second = first.next

        # Swap
        first.next = second.next
        second.next = first
        prev.next = second

        # Move prev
        prev = first

    return dummy.next


# Dry Run: [1,2,3,4]
"""
Initial: dummy → 1 → 2 → 3 → 4
prev = dummy

Iteration 1:
first=1, second=2
1.next = 3
2.next = 1
dummy.next = 2
Result: dummy → 2 → 1 → 3 → 4
prev = 1

Iteration 2:
first=3, second=4
3.next = None
4.next = 3
1.next = 4
Result: dummy → 2 → 1 → 4 → 3

Final: [2,1,4,3]
"""
```

## Q16: Sort Linked List (Merge Sort)

```python
def sort_list(head):
    """Sort linked list using merge sort"""
    if not head or not head.next:
        return head

    # Find middle
    slow = head
    fast = head.next
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    # Split list
    mid = slow.next
    slow.next = None

    # Recursively sort both halves
    left = sort_list(head)
    right = sort_list(mid)

    # Merge sorted halves
    return merge_sorted(left, right)


# Dry Run: [4,2,1,3]
"""
Split: [4,2] and [1,3]
Split again: [4] and [2], [1] and [3]
Merge: [2,4] and [1,3]
Merge: [1,2,3,4]
"""
```

## Q17: Reorder List

```python
def reorder_list(head):
    """Reorder: L0→Ln→L1→Ln-1→L2→Ln-2→..."""
    if not head or not head.next:
        return head

    # Find middle
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    # Reverse second half
    prev = None
    current = slow
    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node

    # Merge two halves alternately
    first = head
    second = prev
    while second.next:
        first_next = first.next
        second_next = second.next

        first.next = second
        second.next = first_next

        first = first_next
        second = second_next

    return head


# Dry Run: [1,2,3,4,5]
"""
Step 1: Find middle = 3
Step 2: Reverse second half [3,4,5] → [5,4,3]
First half: [1,2,3]
Second half: [5,4,3]

Step 3: Merge alternately
1 → 5 → 2 → 4 → 3

Result: [1,5,2,4,3]
"""
```

## Q18: Flatten Multilevel Doubly Linked List

```python
def flatten(head):
    """Flatten multilevel doubly linked list"""
    if not head:
        return None

    current = head
    while current:
        if current.child:
            # Save next
            next_node = current.next

            # Flatten child
            child = flatten(current.child)

            # Connect current to child
            current.next = child
            child.prev = current
            current.child = None

            # Find end of child list
            while current.next:
                current = current.next

            # Connect to next
            current.next = next_node
            if next_node:
                next_node.prev = current

        current = current.next

    return head
```

## Q19: LRU Cache Implementation

```python
class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.cache = {}  # key -> node
        self.head = Node(0, 0)  # dummy head
        self.tail = Node(0, 0)  # dummy tail
        self.head.next = self.tail
        self.tail.prev = self.head

    def _remove_node(self, node):
        """Remove node from list"""
        node.prev.next = node.next
        node.next.prev = node.prev

    def _add_to_front(self, node):
        """Add node right after head"""
        node.next = self.head.next
        node.prev = self.head
        self.head.next.prev = node
        self.head.next = node

    def get(self, key):
        """Get value and mark as recently used"""
        if key in self.cache:
            node = self.cache[key]
            self._remove_node(node)
            self._add_to_front(node)
            return node.value
        return -1

    def put(self, key, value):
        """Add or update key-value pair"""
        if key in self.cache:
            node = self.cache[key]
            node.value = value
            self._remove_node(node)
            self._add_to_front(node)
        else:
            if len(self.cache) >= self.capacity:
                # Remove least recently used (tail's prev)
                lru = self.tail.prev
                self._remove_node(lru)
                del self.cache[lru.key]

            new_node = Node(key, value)
            self.cache[key] = new_node
            self._add_to_front(new_node)


# Dry Run: LRU Cache capacity 2
"""
put(1,1): cache={1:1}, list: [1]
put(2,2): cache={1:1,2:2}, list: [2,1]
get(1):   return 1, list: [1,2]
put(3,3): cache full, remove 2, cache={1:1,3:3}, list: [3,1]
get(2):   return -1 (not found)
"""
```

## Q20: Clone List with Random Pointer

```python
def copy_random_list(head):
    """Clone linked list with random pointers"""
    if not head:
        return None

    # Step 1: Create interleaved copy
    current = head
    while current:
        new_node = Node(current.data)
        new_node.next = current.next
        current.next = new_node
        current = new_node.next

    # Step 2: Set random pointers
    current = head
    while current:
        if current.random:
            current.next.random = current.random.next
        current = current.next.next

    # Step 3: Separate lists
    current = head
    new_head = head.next
    while current:
        copy = current.next
        current.next = copy.next
        if copy.next:
            copy.next = copy.next.next
        current = current.next

    return new_head


# Dry Run: [1,2,3] with random pointers
"""
Original: 1→2→3, 1.random=3, 2.random=1, 3.random=2

Step 1: Interleave
1→1'→2→2'→3→3'

Step 2: Set random
1'.random = 1.random.next = 3'
2'.random = 2.random.next = 1'
3'.random = 3.random.next = 2'

Step 3: Separate
Original: 1→2→3
Copy: 1'→2'→3'
"""
```

## Q21: Add Two Numbers

```python
def add_two_numbers(l1, l2):
    """Add two numbers represented by linked lists"""
    dummy = Node(0)
    current = dummy
    carry = 0

    while l1 or l2 or carry:
        sum_val = carry

        if l1:
            sum_val += l1.data
            l1 = l1.next
        if l2:
            sum_val += l2.data
            l2 = l2.next

        carry = sum_val // 10
        current.next = Node(sum_val % 10)
        current = current.next

    return dummy.next


# Dry Run: l1=[2,4,3] (342), l2=[5,6,4] (465)
"""
Sum = 342 + 465 = 807

Step 1: 2+5=7, carry=0, result=[7]
Step 2: 4+6=10, carry=1, digit=0, result=[7,0]
Step 3: 3+4+1=8, carry=0, digit=8, result=[7,0,8]

Output: [7,0,8] (807)
"""
```

## Q22: Partition List

```python
def partition(head, x):
    """Partition list around value x"""
    before_head = Node(0)
    before = before_head
    after_head = Node(0)
    after = after_head

    current = head
    while current:
        if current.data < x:
            before.next = current
            before = before.next
        else:
            after.next = current
            after = after.next
        current = current.next

    # Connect partitions
    after.next = None
    before.next = after_head.next

    return before_head.next


# Dry Run: [1,4,3,2,5,2], x=3
"""
Before list (less than 3): 1→2→2
After list (≥3): 4→3→5

Result: 1→2→2→4→3→5
"""
```

## Q23: Reverse Nodes in K-Group

```python
def reverse_k_group(head, k):
    """Reverse nodes in groups of k"""
    dummy = Node(0)
    dummy.next = head
    prev_group = dummy

    while True:
        # Check if k nodes remain
        kth = prev_group
        for _ in range(k):
            kth = kth.next
            if not kth:
                return dummy.next

        # Reverse k nodes
        first = prev_group.next
        curr = first
        prev = kth.next

        for _ in range(k):
            next_node = curr.next
            curr.next = prev
            prev = curr
            curr = next_node

        # Connect with previous group
        prev_group.next = prev
        prev_group = first


# Dry Run: [1,2,3,4,5], k=3
"""
First group: [1,2,3] → reverse → [3,2,1]
Remaining: [4,5] (less than k, keep as is)

Result: [3,2,1,4,5]
"""
```
