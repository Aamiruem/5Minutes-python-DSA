# Q1: Create & Display Linked List
Problem: Build a linked list from array and display it.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def create_from_array(self, arr):
        if not arr:
            return None

        self.head = Node(arr[0])
        current = self.head

        for i in range(1, len(arr)):
            current.next = Node(arr[i])
            current = current.next
        return self.head

    def display(self):
        elements = []
        current = self.head
        while current:
            elements.append(current.data)
            current = current.next
        return elements
    
    
```
