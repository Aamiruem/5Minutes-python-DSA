# Q1: Create & Display Linked List
Problem: Build a linked list from array and display it.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def create_from_array(self, arr):
        if not arr:
            return None

        self.head = Node(arr[0])
        current = self.head

        for i in range(1, len(arr)):
            current.next = Node(arr[i])
            current = current.next
        return self.head

    def display(self):
        elements = []
        current = self.head
        while current:
            elements.append(current.data)
            current = current.next
        return elements

    def insert_at_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def delete_from_end(self):
        if not self.head:
            return None

        if not self.head.next:
            deleted_data = self.head.data
            self.head = None
            return deleted_data

        current = self.head
        while current.next.next:
            current = current.next
            deleted_data = current.next.data
            current.next = None

            return deleted_data


# crete linked list from array
ll = LinkedList()
ll.create_from_array([10, 20, 30, 40])
print("Original List: ", ll.display())

# Insert at end
ll.insert_at_end(50)
print("After inserting 50: ", ll.display())

# Delete from end
deleted = ll.delete_from_end()
print("delete element:", deleted)
print("List after deletion:", ll.display())

print("just deletion ", ll.display())


# delete until empty
ll.delete_from_end()
ll.delete_from_end()
ll.delete_from_end()
ll.delete_from_end()
print("List after deleting all:", ll.display())
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def create_from_array(self, arr):
        if not arr:
            return None

        self.head = Node(arr[0])
        current = self.head

        for i in range(1, len(arr)):
            current.next = Node(arr[i])
            current = current.next
        return self.head

    def display(self):
        elements = []
        current = self.head
        while current:
            elements.append(current.data)
            current = current.next
        return elements

    def insert_at_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def delete_from_end(self):
        if not self.head:  # empty list
            return None

        if not self.head.next:  # only one node
            deleted_data = self.head.data
            self.head = None
            return deleted_data

        current = self.head
        while current.next.next:  # stop at 2nd last node
            current = current.next
        deleted_data = current.next.data
        current.next = None
        return deleted_data
```

```python
# Create linked list from array
ll = LinkedList()
ll.create_from_array([10, 20, 30, 40])
print("Original List:", ll.display())

# Insert at end
ll.insert_at_end(50)
print("After inserting 50:", ll.display())

# Delete from end
deleted = ll.delete_from_end()
print("Deleted element:", deleted)
print("List after deletion:", ll.display())

# Delete until empty
ll.delete_from_end()
ll.delete_from_end()
ll.delete_from_end()
ll.delete_from_end()
print("List after deleting all:", ll.display())
```

# Q2: Count Nodes

```python
def count_nodes(head):
    count = 0
    current = head
    while current:
        count += 1
        current = current.next
    return count


def count_node_recuresive(head):
    if not head:
        return 0
    return 1 + count_node_recuresive(head.next)
```

# Q3: Search Element

```python
def search_elements(head, target):
    current = head
    position = 0

    while current:
        if current.data == target:
            return position
        current = current.next
        position += 1

        return -1
```
