# Q2: Count Nodes

```python
def count_nodes(head):
    count = 0
    current = head
    while current:
        count += 1
        current = current.next
    return count


def count_node_recuresive(head):
    if not head:
        return 0
    return 1 + count_node_recuresive(head.next)
```

```python
# Method 1: Iterative (3 Pointers)
def reverse_iterative(head):
    """Reverse using 3 pointers"""
    prev = None
    current = head

    while current:
        next_node = current.next  # Save next
        current.next = prev  # Reverse link
        prev = current  # Move prev
        current = next_node  # Move current

    return prev


# Dry Run: [1] → [2] → [3] → None
"""
Initial: prev=None, current=1
┌─────┐    ┌─────┐    ┌─────┐
│  1  │───→│  2  │───→│  3  │───→ None
└─────┘    └─────┘    └─────┘
  ↑
current

Iteration 1:
next_node = 2 (save)
1.next = None (reverse)
prev = 1, current = 2

None ← [1]    [2] → [3] → None
        ↑      ↑
       prev  current

Iteration 2:
next_node = 3
2.next = 1
prev = 2, current = 3

None ← [1] ← [2]    [3] → None
              ↑      ↑
             prev  current

Iteration 3:
next_node = None
3.next = 2
prev = 3, current = None

None ← [1] ← [2] ← [3]
                    ↑
                   prev

Return prev (3)

Result: [3] → [2] → [1] → None
"""


# Method 2: Recursive
def reverse_recursive(head):
    """Reverse recursively"""
    # Base case
    if not head or not head.next:
        return head

    # Reverse rest of list
    new_head = reverse_recursive(head.next)

    # Reverse current node
    head.next.next = head
    head.next = None

    return new_head


# Dry Run: [1] → [2] → [3]
"""
Call 1: reverse(1)
  Call 2: reverse(2)
    Call 3: reverse(3) → returns 3 (base case)
  
  Unwind Call 2: 
  2.next.next = 2 → 3.next = 2
  2.next = None
  Returns 3

Unwind Call 1:
1.next.next = 1 → 2.next = 1
1.next = None
Returns 3

Result: 3 → 2 → 1 → None
"""


# Method 3: Stack
def reverse_stack(head):
    """Reverse using stack"""
    if not head:
        return None

    stack = []
    current = head
    while current:
        stack.append(current)
        current = current.next

    head = stack.pop()
    current = head
    while stack:
        current.next = stack.pop()
        current = current.next
    current.next = None

    return head


# Dry Run: [1] → [2] → [3]
"""
Push all to stack:
stack = [1, 2, 3] (top is 3)

Pop 3 → head = 3
Pop 2 → 3.next = 2
Pop 1 → 2.next = 1
Set 1.next = None

Result: [3] → [2] → [1] → None
"""
```

## Q7: Find Middle Element

```python
def find_middle(head):
    """Find middle using slow/fast pointers"""
    if not head:
        return None

    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    return slow.data


# Dry Run: [1] → [2] → [3] → [4] → [5]
"""
Initial: slow=1, fast=1

Step 1: slow=2, fast=3
Step 2: slow=3, fast=5
fast.next is None, stop

Middle = 3

For even list: [1] → [2] → [3] → [4]
Step 1: slow=2, fast=3
Step 2: slow=3, fast=None
fast is None, stop

Middle = 3 (second middle)
"""
```

## Q8: Find Nth Node from End

```python
def nth_from_end(head, n):
    """Find nth node from end using gap method"""
    if not head or n <= 0:
        return None

    first = head
    second = head

    # Create gap of n nodes
    for _ in range(n):
        if not first:
            return None  # n is larger than list
        first = first.next

    # Move both until first reaches end
    while first:
        first = first.next
        second = second.next

    return second.data


# Dry Run: [1] → [2] → [3] → [4] → [5], n=2 (find 2nd from end)
"""
Step 1: Create gap
first moves 2 steps: first at 3
second at 1

[1] → [2] → [3] → [4] → [5]
 ↑            ↑
second      first

Step 2: Move both
first=4, second=2
first=5, second=3
first=None, second=4, stop

Return 4 (2nd from end)
"""
```

## Q9: Detect and Remove Cycle

```python
def detect_cycle(head):
    """Floyd's Cycle Detection"""
    if not head or not head.next:
        return None

    slow = head
    fast = head

    # Detect cycle
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return slow  # Meeting point

    return None


def remove_cycle(head):
    """Detect and remove cycle"""
    # Detect cycle
    meeting_point = detect_cycle(head)
    if not meeting_point:
        return False

    # Find cycle start
    slow = head
    while slow != meeting_point:
        slow = slow.next
        meeting_point = meeting_point.next

    cycle_start = slow

    # Find last node in cycle
    current = cycle_start
    while current.next != cycle_start:
        current = current.next

    # Break cycle
    current.next = None
    return True


# Dry Run: [1] → [2] → [3] → [4] → [5] → [3] (cycle)
"""
Cycle Detection:
Step 1: slow=2, fast=3
Step 2: slow=3, fast=5
Step 3: slow=4, fast=4 (meet!)

Cycle Start:
slow=head(1), meeting=4
Step 1: slow=2, meeting=5
Step 2: slow=3, meeting=3 (meet at cycle start)

Remove:
Find last node (5)
5.next = None

Result: [1] → [2] → [3] → [4] → [5] → None
"""
```

## Q10: Merge Two Sorted Lists

```python
def merge_sorted(l1, l2):
    """Merge two sorted linked lists"""
    dummy = Node(0)
    current = dummy

    while l1 and l2:
        if l1.data <= l2.data:
            current.next = l1
            l1 = l1.next
        else:
            current.next = l2
            l2 = l2.next
        current = current.next

    # Attach remaining
    if l1:
        current.next = l1
    if l2:
        current.next = l2

    return dummy.next


# Dry Run: l1=[1,3,5], l2=[2,4,6]
"""
dummy → None
current = dummy

Step 1: 1≤2 → current.next=1, l1=3
dummy → 1

Step 2: 3≤2? No → current.next=2, l2=4
dummy → 1 → 2

Step 3: 3≤4 → current.next=3, l1=5
dummy → 1 → 2 → 3

Step 4: 5≤4? No → current.next=4, l2=6
dummy → 1 → 2 → 3 → 4

Step 5: 5≤6 → current.next=5, l1=None
dummy → 1 → 2 → 3 → 4 → 5

l1 is None, attach l2
dummy → 1 → 2 → 3 → 4 → 5 → 6

Result: [1,2,3,4,5,6]
"""
```

## Q11: Palindrome Linked List

```python
def is_palindrome(head):
    """Check if linked list is palindrome"""
    if not head or not head.next:
        return True

    # Find middle
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    # Reverse second half
    prev = None
    while slow:
        next_node = slow.next
        slow.next = prev
        prev = slow
        slow = next_node

    # Compare first and second half
    first = head
    second = prev
    while second:
        if first.data != second.data:
            return False
        first = first.next
        second = second.next

    return True


# Dry Run: [1,2,3,2,1]
"""
Step 1: Find middle
slow=1, fast=1
slow=2, fast=3
slow=3, fast=5 (fast.next is None, stop)
middle = 3

Step 2: Reverse second half [3,2,1] → [1,2,3]
Original: 1 → 2 → 3 → 2 → 1
After reverse: 1 → 2 → 3 → 1 → 2

Step 3: Compare
1==1 ✓
2==2 ✓
3==3 ✓ (stop at None)

Return True
"""
```

## Q12: Intersection of Two Linked Lists

```python
def get_intersection(headA, headB):
    """Find intersection point of two linked lists"""
    if not headA or not headB:
        return None
    
    # Get lengths
    lenA, lenB = 0, 0
    currA, currB = headA, headB
    
    while currA:
        lenA += 1
        currA = currA.next
    
    while currB:
        lenB += 1
        currB = currB.next
    
    # Reset pointers
    currA, currB = headA, headB
    
    # Move longer list pointer
    if lenA > lenB:
        for _ in range(lenA - lenB):
            currA = currA.next
    else:
        for _ in range(lenB - lenA):
            currB = currB.next
    
    # Find intersection
    while currA and currB:
        if currA == currB:
            return currA
        currA = currA.next
        currB = currB.next
    
    return None

# Dry Run:
"""
List A: 1 → 2 → 3 → 4 → 5
List B: 6 → 4 → 5 (4 is intersection)

lenA = 5, lenB = 3
Move A by 2: A at 3, B at 6

Compare:
3≠6, move both
4=4 ✓ (intersection found)
"""
```
