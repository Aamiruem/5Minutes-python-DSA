# Linear Search

```python
arr = [2, 3, 4, 67, 45, 34, 23, 98]

target = 45
found = False
for i in range(len(arr)):
    if arr[i] ==  target:
        print(f"{arr[i]} Element found at index {i}")
        found = True
        
        break
    
if not found:
    print("Element not found")
```
