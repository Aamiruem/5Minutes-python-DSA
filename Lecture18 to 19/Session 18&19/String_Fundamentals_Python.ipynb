## 1. Strings in Python are Immutable

```python
s = "hello"
try:
    s[0] = 'H'  
except TypeError as e:
    print("Error:", e)

s = 'H' + s[1:]
print("Modified string:", s)
```

```python
s = "hello"                # Start with a string "hello"
try:
    s[0] = 'H'             # Try to change the first letter directly
except TypeError as e:     
    print("Error:", e)     # Strings cannot be changed directly → error

s = 'H' + s[1:]            # Make a new string: 'H' + "ello" = "Hello"
print("Modified string:", s)
```

```python
# explanation
# In Python, strings are immutable → you cannot change them directly like s[0] = 'H'.

# That’s why the first part gives a TypeError.

# To “change” a string, you actually build a new string:

# 'H' + s[1:]

# s[1:] means “everything from index 1 onwards” → "ello".

# So 'H' + "ello" = "Hello".

# Dry run
# s = "hello"

# Try s[0] = 'H' → ❌ Error: 'str' object does not support item assignment

# Catch error → print:

# Code
# Error: 'str' object does not support item assignment
# Build new string: 'H' + s[1:] → 'H' + "ello" → "Hello"

# Print:

# Code
# Modified string: Hello
```

## 2. Creating and Accessing Strings

```python
s = "Python Programming"
print("First character:", s[0])
print("Last character:", s[-1])
print("Length of string:", len(s))
```

## 3. Slicing and Indexing

```python
s = "Hello, World!"
print("s[0:5]:", s[0:5])  
print("s[:5]:", s[:5])    
print("s[7:]:", s[7:])    
print("s[::-1]:", s[::-1])  
```

## 4. String Methods

```python
s = "  Hello, Python!  "
print("Original:", repr(s))
print("lower():", s.lower())
print("upper():", s.upper())
print("strip():", s.strip())

words = s.strip().split()
print("split():", words)

joined = "-".join(words)
print("join():", joined)

replaced = s.replace("Python", "World")
print("replace():", replaced)
```

## 5. String Formatting
### a. f-strings (Python 3.6+ syntax)

```python
name = "shridhar"
age = 30
print(f"My name is {name} and I am {age} years old.")
```

### b. .format() Method

```python
name = "shridhar"
language = "Python"
print("Hello, my name is {} and I code in {}.".format(name, language))
```
