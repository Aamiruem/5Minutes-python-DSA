```python
# checking for anagram
from collections import Counter

s1 = "listen"
s2 = "silent"

is_anagram = Counter(s1) == Counter(s2)

print(f"'{s1}', and '{s2}', are anagrams: ", is_anagram)
```

```python
# Split an apple → cut it into two halves.

# Split the bill → share the cost among people.

# Split a string in Python → break text into smaller words.


# Checking substring
s = "Hello Technology"
rev_str = s[::-1]

print("Reversed String:", rev_str)

words = s.split()  # split means = toda
reversed_words = [word[::-1] for word in words] # har ek word par jakar reversed kiya
print("Rversed words:", " ".join(reversed_words)) # then joda reversed word ko
```

```python
# checking palindrome

s = "madams"  # after reversed word no change looks similar as like original
p = s == s[::-1]
print(f"'{s}' is a palindrome:", p)
```

```python
# checking anagram => different word but same character
from collections import Counter

s1 = "listen"
s2 = "silent"
is_anagram = Counter(s1) == Counter(s2)

print(f"'{s1}' and '{s2}' are anagrams:", is_anagram)
```

```python
# Character Frequency Count using Dictionary
s = "programming"
freq = {}  # => store the value in dictionary

for char in s:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1
print("Character frequencies")
for char, count in freq.items():
    print(f"{char}: {count}")
```

```python
# Removing Duplicates from a string

s = "programming"

result = " "
seen = set()

for char in s:
    if char not in seen:
        seen.add(char)
        result += char

print("String after removing duplicates:", result)
```
