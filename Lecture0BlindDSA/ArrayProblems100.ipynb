### Q2. Write a program to find the sum of all elements in an array.

```python
def array_sum(arr):
    total = 0
    for num in arr:
        total += num
    return total


# Test
arr = [1, 2, 3, 4, 5]
print(f"Array: {arr}")
print(f"Sum: {array_sum(arr)}")

# Dry Run:
# arr = [1, 2, 3, 4, 5]
# Step 1: total = 0
# Step 2: num=1 → total = 0+1 = 1
# Step 3: num=2 → total = 1+2 = 3
# Step 4: num=3 → total = 3+3 = 6
# Step 5: num=4 → total = 6+4 = 10
# Step 6: num=5 → total = 10+5 = 15
# Output: 15
```
