```python
class student:
    name = "Aamir"

    def study(self):
        print(f"{self.name} is studying")


student1 = student()
print(student1.name)
student1.study()
```

```python
class student:
    def __init__(self, name, age):
        self.name = name
        self.age = age


student1 = student("Aamir", 20)
student2 = student("Ali", 22)
print(student1.name, student1.age)
print(student2.name, student2.age)
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show(self):
        print(f"Name: {self.name}, Salary: {self.salary}")


employee1 = Employee("Aamir", 50000)
employee2 = Employee("Ali", 60000)

employee1.show()
employee2.show()
```

```python
class student:
    def __init__(self, name):
        self.name = name


student1 = student("Aamir")
student2 = student("Ali")

print(student1.name)
print(student2.name)

student1.name = "Aamir Khan"
print(student1.name)


student2.name = "Azad"
print(student2.name)
```

```python
class student:
    college = "UEM Jaipur"

    def __init__(self, name):
        self.name = name


student1 = student("Aamir")
student2 = student("Ali")

print(student.college, "\n")
print(student1.name, student1.college, "\n")
print(student2.name, student2.college, "\n")
```
