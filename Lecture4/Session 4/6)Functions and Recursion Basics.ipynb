### Variable-Length Arguments

```python
def sum_all(*args):  # star player act as list of arguments, it can take any number of arguments ( throw out the multiple ball nad catch the star player)
    return sum(args) # send the list of arguments to sum function which will return the sum of all the arguments

print(sum_all(1, 2, 3, 4))
```

## 4. Scope: Local vs Global Variables

```python
count = 0 # bahri duniya variable

def increment():
    count1 = 5 # function ke andar ka variable
    count1 += 1 # 
    print(count1)

increment() 
print('Count:', count)
```

```python
count = 0

def increment():
    count = 5
    count += 1
    print(count)

increment()
print('Count:', count)
```

```python
count = 0

def increment():
    global count
    count += 1

increment()
print('Count:', count)
```

## 5. Recursion Fundamentals and Base Case

```python
def countdown(n):
    if n == 0: # base case jab 0 ho jaye to recursion ko rok do aur return kar do GOOO!
        print('Gooo!')
    else:
        print(n) 
        countdown(n - 1)

countdown(3)
```

## 6. Simple Recursion Examples

### Factorial

```python
def fact(n):
    print(n) # 
    if n == 0:
        return 1
    else:
        return n * fact(n - 1)


fact(5)
```

```python
def factorial(n):
    print(n)
    return 1 if n == 0 else n * factorial(n - 1)
factorial(3)
```

### Fibonacci (nth number)

```python
def fib(n):
    if n <= 1:
        return n
    else:
        return fib(n - 1) + fib(n - 2)
fib(5)
```

if n <= 1: → Base case. If n is 0 or 1, just return n (because Fibonacci of 0 = 0, Fibonacci of 1 = 1).

else: → Recursive case. If n is bigger than 1, calculate:

### Sum of Digits

```python
def sum_digits(n):
    print(n)
    return 0 if n == 0 else n % 10 + sum_digits(n // 10)
sum_digits(12)
```
