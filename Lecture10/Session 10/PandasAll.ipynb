```python
data = {
    "id": [101, 102, 103, 104, 105, 106],
    "name": ["Aamir", "Sara", "Ravi", "Neha", "Kabir", "Sara"],
    "department": ["AI", "HR", "AI", "Sales", "Sales", "HR"],
    "city": ["Delhi", "Jaipur", "Delhi", "Mumbai", "Jaipur", "Jaipur"],
    "salary": [60000, 45000, 70000, None, 52000, 45000],
    "experience": [2, 3, 5, 1, 4, 3],
    "rating": [4, 5, 4, 3, 5, 5],
    "joining_date": [
        "2024-01-10",
        "2023-05-15",
        "2021-07-20",
        "2025-02-01",
        "2022-09-12",
        "2023-05-15"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.head())        # First 5 rows
print(df.head(2))       # First 2 rows
print(df.tail())        # Last 5 rows

print(df.shape)         # Number of rows and columns
print(df.columns)       # Column names
print(df.index)         # Row indexes
print(df.dtypes)        # Data type of every column
print(df.info())        # Complete structure
```

```python
print(df.describe())
```

```python
print(df.describe(include="all"))
```

```python
print(df.size)          # Total cells
print(df.ndim)          # Number of dimensions
print(df.empty)         # Is DataFrame empty?
print(df.values)        # Data as NumPy array
```

```python
names = df["name"]
print(names)
```

```python
result = df[["name", "salary", "department"]]
print(result)
```

```python
# df.loc[row_labels, column_labels]

print(df.loc[0])

print(df.loc[0, "name"])

print(df.loc[0:2, ["name", "salary"]])
```

```python
# df.iloc[row_positions, column_positions]


print(df.iloc[0])          # First row
print(df.iloc[0, 1])       # First row, second column
print(df.iloc[0:3, 1:4])   # Rows 0–2 and columns 1–3
```

```python
print(df.at[0, "name"])   # Label-based
print(df.iat[0, 1])       # Position-based
```
