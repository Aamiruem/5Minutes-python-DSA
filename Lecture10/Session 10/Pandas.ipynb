```python
# 🧩 isnull()
# Purpose: Checks for missing values (NaN) in your DataFrame.

# Returns: A DataFrame of True/False values.


import pandas as pd

data = {"Name": ["Aamir", "Ravi", None], "Age": [24, None, 28]}
df = pd.DataFrame(data)

print(df.isnull())
```

```python
# 🧩 fillna()
# Purpose: Fills missing values (NaN) with a specified value or method.

# Common uses:

# Replace with a fixed value:

df.fillna(0)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df.fillna({"Age": df["Age"].mean()}, inplace=True)
```

```python
import pandas as pd

data = {"Name": ["Aamir", "Ravi", None], "Age": [24, None, 28]}
df = pd.DataFrame(data)

# Safe way
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Name"] = df["Name"].fillna("Unknown")

print(df)
```

```python
# df.fillna(method="ffill")
df.ffill()
```

```python
df.bfill()
```

```python
# Backward fill
df = df.bfill()

# Forward fill
df = df.ffill()
```

```python
import pandas as pd

data = {"Name": ["Aamir", None, "Neha"], "Age": [24, None, 28]}
df = pd.DataFrame(data)

# Safe backward fill
df = df.bfill()

print(df)
```

```python
df["Age"] = df["Age"].bfill()
```

```python
df["Name"] = df["Name"].fillna("Unknown")
df["Age"]  = df["Age"].fillna(df["Age"].mean())
```

```python
df.fillna({"Name": "Unknown", "Age": df["Age"].mean()}, inplace=True)
```
