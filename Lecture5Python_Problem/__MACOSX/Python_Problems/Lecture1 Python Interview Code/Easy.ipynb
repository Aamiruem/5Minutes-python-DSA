```python
# counting special character
def count_special_characters(string):
    spec_char = "!@#$%^&*()<>?/\[]{};:~`+=-_"
    count = 0
    for i in string:
        if i in spec_char:
            count += 1
    return count
text = "Hello! How are you? I hope you're doing well. Have a great day! @#$%^&*()"
special_char_count = count_special_characters(text)
print("Number of special characters:", special_char_count)


# Word‑by‑Word Meaning
# def count_special_characters(string): → define a function that takes a string as input.

# spec_char = "...symbols..." → list of special characters you want to check.

# count = 0 → start a counter at zero.

# for i in string: → loop through each character in the text.

# if i in spec_char: → check if the character is inside the special character list.

# count += 1 → if yes, increase the counter by 1.

# return count → give back the total count.

# What happens
# Input text = "Hello! How are you? I hope you're doing well. Have a great day! @#$%^&*()"

# The function checks each character.

# Special characters found: ! ? ! @ # $ % ^ & * ( )

# Total count = 13

# 👉 Output will be:

# Code
# Number of special characters: 13
# 💡 Easy way to remember:

# Loop → check each character.

# if in spec_char → only count special ones.

# += 1 → add to total.
```

```python
import re
string = "c o d e"
space = re.compile(r'\s+')
result = re.sub(space, '', string)

print(result)



# Word‑by‑Word Meaning
# import re → bring in Python’s regular expression (regex) module.

# string = "c o d e" → make a variable called string with the text "c o d e" (spaces between letters).

# space = re.compile(r'\s+') →

# re.compile(...) → prepare a regex pattern.

# r'\s+' → means “one or more whitespace characters” (like spaces, tabs, newlines).

# So space is now a pattern that matches spaces.

# result = re.sub(space, '', string) →

# re.sub(pattern, replacement, text) → replace all matches of the pattern with something else.

# space → the pattern (spaces).

# '' → replace with nothing (remove).

# string → the text to clean.

# So this removes all spaces from "c o d e".

# print(result) → show the final result.

# What happens
# Original string = "c o d e"

# Regex finds spaces (" ").

# Replace them with nothing.

# Final result = "code"

# 👉 Output will be:

# Code
# code
# 💡 Easy way to remember:

# \s = whitespace (space, tab, newline).

# + = one or more.

# re.sub = search and replace.
```

```python
def pyramid(n):
    for i in range(n):
        # print spaces
        for j in range(n - i - 1):
            print(" ", end="")
        # print stars
        for j in range(2 * i + 1):
            print("*", end="")
        # move to next line
        print()

pyramid(5)
# explanation
# Outer loop for i in range(n) → controls the number of rows.

# First inner loop for j in range(n - i - 1) → prints spaces to align the pyramid.

# Second inner loop for j in range(2 * i + 1) → prints stars (*).

# print() at the end → moves to the next line after each row.
```

# 🔍 Explanation


def diamond(n):

Defines a function named diamond that takes one argument n (the height of the diamond).

for i in range(n):

This loop builds the upper pyramid (top half of the diamond).

i goes from 0 to n-1.

print(" " * (n - i - 1) + "*" * (2 * i + 1))

" " * (n - i - 1) → prints spaces to center the stars.

Example: when i=0 and n=5, it prints 4 spaces.

"*" * (2 * i + 1) → prints stars.

Formula ensures odd numbers: 1, 3, 5, 7, …

Together, spaces + stars form each row of the pyramid.

for i in range(n - 2, -1, -1):

This loop builds the lower inverted pyramid (bottom half of the diamond).

Starts from n-2 (second-to-last row) down to 0.

This avoids repeating the middle row twice.

print(" " * (n - i - 1) + "*" * (2 * i + 1))

Same logic as above, but now printing rows in reverse order.

This creates the bottom half that connects seamlessly with the top half.

diamond(5)

Calls the function with n=5.

Prints a diamond of height 5.

```python
def diamond(n):
    # upper pyramid
    for i in range(n):
        print(" " * (n - i - 1) + "*" * (2 * i + 1))
    # lower inverted pyramid
    for i in range(n - 2, -1, -1):
        print(" " * (n - i - 1) + "*" * (2 * i + 1))

# Example: diamond of height 5
diamond(5)
```
