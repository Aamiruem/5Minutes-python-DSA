```python
def missing_numbers(numbers):
    n = len(numbers)

    expected_sum = n * (n + 1) / 2
    actual_sum = sum(numbers)

    return expected_sum - actual_sum


print(missing_numbers([3, 0, 1]))
```

```python
def maximum_window_sum(numbers, k):
    if k <= 0 or k > len(numbers):
        raise ValueError("Invalid window size")

    window_sum = sum(numbers[:k])
    maximum_sum = window_sum

    for right in range(k, len(numbers)):
        left = right - k

        window_sum += numbers[right]
        window_sum -= numbers[left]

        maximum_sum = max(maximum_sum, window_sum)

    return maximum_sum


print(maximum_window_sum([2, 1, 5, 1, 3, 2], 3))
```

```python
def find_duplicate(numbers):
    seen = set()

    for number in numbers:
        if number in seen:
            return number
        seen.add(number)

    return None


print(find_duplicate([1, 3, 4, 2, 6, 2, 3, 5, 3, 4]))
```

```python
def find_duplicates(numbers):
    seen = set()
    duplicates = set()

    for number in numbers:

        if number in seen:
            duplicates.add(number)
        else:
            seen.add(number)
    return list(duplicates)


print(find_duplicates([1, 3, 4, 2, 6, 2, 3, 5, 3, 4]))
```

```python
word = "programming"

# Step 1: Create dictionary
char_count = {}

# Step 2: Count characters
for ch in word:
    if ch in char_count:
        char_count[ch] += 1
    else:
        char_count[ch] = 1

# Step 3: Print repeated characters
for ch, count in char_count.items():
    if count > 1:
        print(f"'{ch}' is repeated {count} times")
```

```python
def majority_element(numbers):
    majority_element = None

    count = 0
    for number in numbers:
        if count == 0:
            majority_element = number

            if number == majority_element:
                count += 1
            else:
                count -= 1

    if numbers.count(majority_element) > len(numbers) // 2:
        return majority_element
    return None


print(majority_element([2, 2, 1, 1, 1, 2, 2]))
```

```python
def count_frequency(numbers):
    freq = {}
    for num in numbers:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1
    return freq


numbers = [2, 2, 1, 1, 1, 2, 2]
result = count_frequency(numbers)


for num in result:
    print(f"{num} appears {result[num]} times")
```

```python
def sort_zero_one_two(numbers):
    low = 0
    middle = 0
    high = len(numbers) - 1

    while middle <= high:
        if numbers[middle] == 0:
            numbers[low], numbers[middle] = numbers[middle], numbers[low]
            low += 1
            middle += 1

        elif numbers[middle] == 1:
            middle += 1

        else:
            numbers[middle], numbers[high] = numbers[middle]
            high -= 1
    return numbers


print(sort_zero_one_two([0, 1, 2, 1, 0, 2, 1, 0]))
```
