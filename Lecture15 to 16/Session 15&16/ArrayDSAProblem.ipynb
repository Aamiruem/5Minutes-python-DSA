```python
def range_sum(prefix, left, right):
    # This function calculates the sum of elements between indices 'left' and 'right'
    # using the prefix sum array.
    # Formula: prefix[right+1] - prefix[left]
    return prefix[right+1] - prefix[left]


numbers = [2, 4, 6, 8, 10]  
# Original list of numbers we want to work with.

prefix = create_prefix_sum(numbers)  
# Create the prefix sum array from 'numbers'.
# Result will be: [0, 2, 6, 12, 20, 30]

print(range_sum(prefix, 1, 3))  # Output: 18 (4 + 6 + 8)
# Explanation:
# prefix[4] - prefix[1] = 20 - 2 = 18
# This gives the sum of numbers[1] to numbers[3] → 4 + 6 + 8 = 18


print(range_sum(prefix, 0, 2))  # 2 + 4 + 6 = 12
# prefix[3] - prefix[0] = 12 - 0 = 12
# Sum of numbers[0] to numbers[2] → 2 + 4 + 6 = 12

print(range_sum(prefix, 2, 4))  # 6 + 8 + 10 = 24
# prefix[5] - prefix[2] = 30 - 6 = 24
# Sum of numbers[2] to numbers[4] → 6 + 8 + 10 = 24

print(range_sum(prefix, 0, 4))  # 2 + 4 + 6 + 8 + 10 = 30
# prefix[5] - prefix[0] = 30 - 0 = 30
# Sum of numbers[0] to numbers[4] → 2 + 4 + 6 + 8 + 10 = 30
```

```python
def find_missing_number(numbers):
    n = len(numbers)  
    # Step 1: Find how many numbers are in the list.
    # Example: numbers = [3, 0, 1] → n = 3

    expected_sum = n * (n + 1) // 2  
    # Step 2: Calculate the sum of numbers from 0 to n using formula n*(n+1)//2.
    # For n = 3 → expected_sum = 3 * 4 // 2 = 6
    # This is the sum if no number was missing.

    actual_sum = sum(numbers)  
    # Step 3: Calculate the sum of the given list.
    # sum([3, 0, 1]) = 4

    return expected_sum - actual_sum  
    # Step 4: The missing number = expected_sum - actual_sum
    # → 6 - 4 = 2
```

```python
def find_missing_using_xor(numbers):
    result = len(numbers)
    
    for index, number in enumerate(numbers):
        result ^= index
        result ^= number
        
    return result

print(find_missing_using_xor([3, 0, 1]))
```

⚡ Why XOR works
XOR has a property: a ^ a = 0 and a ^ 0 = a.

If you XOR all indices and all numbers together, the duplicates cancel out.

The only value left is the missing number.

```python
def find_missing_using_xor(numbers):
    result = len(numbers)  
    # Step 1: Initialize result with n (length of the list).
    # If numbers has length 3, then n = 3.
    # This accounts for the "last index" that isn't included in enumerate.

    for index, number in enumerate(numbers):
        # Step 2: Loop through each index and number in the list.
        
        result ^= index  
        # XOR with the index.
        
        result ^= number  
        # XOR with the actual number in the list.

    return result  
    # Step 3: After the loop, result will hold the missing number.
```

```python
def find_duplicate(numbers):
    
    
    seen = set()
    
    for number in numbers:
        if number in seen:
            return number
        
        seen.add(number)
        
    return None

    
print(find_duplicate([1, 3, 4, 2, 2]))  # Output: 2
```

```python
def find_all_duplicates(numbers):
    seen = set()          # keeps track of numbers we've seen
    duplicates = set()    # keeps track of duplicates found
    
    for number in numbers:
        if number in seen:
            duplicates.add(number)   # add to duplicates instead of returning immediately
        else:
            seen.add(number)
    
    return list(duplicates)   # return all duplicates as a list

print("These are the duplicate values: ", find_all_duplicates([1, 3, 4, 4, 2, 2, 3]))
```

```python
def majority_element(numbers):
    duplicate = None        # Step 1: Initialize duplicate as None
    count = 0               # Step 2: Initialize count as 0

    for number in numbers:  # Step 3: Loop through each number in the list
        if count == 0:      # If count is 0, choose current number as new duplicate
            duplicate = number
        if number == duplicate:  
            count += 1      # If current number matches duplicate, increase count
        else:
            count -= 1      # If current number is different, decrease count

    return duplicate        # Step 4: Return the duplicate (majority element)


numbers = [2, 2, 1, 1, 1, 2, 3, 2, 3, 2, 6, 1, 1]

print(majority_element(numbers))
```

```python
def merge_sorted_list(list1, list2):
    f = 0 # first pointer for list1
    s = 0 # second pointer for list2
    
    merged = [] # list to store the merged result
    
    while f<len(list1) and s < len(list2):
        if(list1[f] <= list2[s]):
            merged.append(list1[f])
            f += 1
        else:
            merged.append(list2[s])
            s += 1
            
    merged.extend(list1[f:])
    merged.extend(list2[s:])


    return merged


list1 = [1, 3, 5, 7]
list2 = [2, 4, 6, 8]


print(merge_sorted_list(list1, list2))  # Output: [1, 2, 3, 4, 5, 6, 7, 8]
```

```python
def product_except_self(numbers):
    length = len(numbers)
    
    result = [1] * length
    prefix_product = 1

    for index in range(length):
       result[index] = prefix_product
       prefix_product *= numbers[index]

    suffix_product = 1
    
    for index in range(length - 1, -1, -1):
        result[index]  *= suffix_product
        suffix_product *= numbers[index]
        
    return result


print(product_except_self([1, 2, 3, 4]))  # Output: [24, 12, 8, 6]
       
       
#        Super simple meaning
# At index 0 → product of all except 1 → 2×3×4 = 24

# At index 1 → product of all except 2 → 1×3×4 = 12

# At index 2 → product of all except 3 → 1×2×4 = 8

# At index 3 → product of all except 4 → 1×2×3 = 6



# Easy dry run with [1, 2, 3, 4]
# First loop (prefix):

# Index 0 → result[0] = 1 → prefix_product = 1×1 = 1

# Index 1 → result[1] = 1 → prefix_product = 1×2 = 2

# Index 2 → result[2] = 2 → prefix_product = 2×3 = 6

# Index 3 → result[3] = 6 → prefix_product = 6×4 = 24
# 👉 After first loop: result = [1, 1, 2, 6]

# Second loop (suffix):

# Index 3 → result[3] = 6×1 = 6 → suffix_product = 1×4 = 4

# Index 2 → result[2] = 2×4 = 8 → suffix_product = 4×3 = 12

# Index 1 → result[1] = 1×12 = 12 → suffix_product = 12×2 = 24

# Index 0 → result[0] = 1×24 = 24 → suffix_product = 24×1 = 24
# 👉 After second loop: result = [24, 12, 8, 6]
```

```python
def product_except_self(numbers):
    length = len(numbers)          # How many numbers are in the list
    result = [1] * length          # Start with a list of 1s
    prefix_product = 1             # Product of numbers before current index

    # First loop: go left to right
    for index in range(length):
       result[index] = prefix_product      # Save product of numbers before this index
       prefix_product *= numbers[index]    # Update prefix product

    suffix_product = 1             # Product of numbers after current index

    # Second loop: go right to left
    for index in range(length - 1, -1, -1):
        result[index] *= suffix_product    # Multiply with product of numbers after this index
        suffix_product *= numbers[index]   # Update suffix product

    return result                  # Return the final list
```

```python
def merge_intervals(intervals):
    if not intervals:
        return []
    intervals.sort(key=lambda interval: interval[0])     
    merged = [intervals[0]]
    
    for current_start, current_end in intervals[1:]:
        previous_end = merged[-1][1]
        
        if current_start <=previous_end:
            merged[-1][1] = max(previous_end, current_end)
        else:
            merged.append([current_start, current_end])
            
    return merged



intervals = [[1, 3], [2, 6], [8, 10], [15, 18]]
print(merge_intervals(intervals))  # Output: [[1, 6], [8, 10], [15, 18]]
            
```

```python
def rotate_matrix_90(matrix):
    size = len(matrix)
    
    for row in range(size):
        for column in range(row + 1, size):
            matrix[row][column], matrix[column][row] = (
                matrix[column][row],
                matrix[row][column]
            )
    for row in matrix:
        row.reverse()
    return matrix

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(rotate_matrix_90(matrix))
```

```python
def set_matrix_zeros(matrix):
    zero_rows = set()
    zero_columns = set()
    
    
    for row in range(len(matrix)):
        for column in range(len(matrix[0])):
            if matrix[row][column] == 0:
                zero_rows.add(row)
                zero_columns.add(column)
    
    for row in range(len(matrix)):
        for column in range(len(matrix[0])):
            if row in zero_rows or column in zero_columns:
                matrix[row][column] = 0
                
                
                
    return matrix



matrix = [
    [1, 1, 1],
    [1, 0, 1],
    [1, 1, 1]
]
result = set_matrix_zeros(matrix)
for row in result:
    print(" ".join(map(str, row)))

print("\nwithout matrix form")
print(set_matrix_zeros(matrix))
```
