## Shallow Copy vs Deep Copy

Shallow copy → Copies only the outer list. If the list has inner lists, they are still shared between the original and the copy.

Deep copy → Copies everything, including inner lists. So the new list is completely independent.

```python
# Example with Shallow Copy
# python
# import copy

list1 = [[10, 20], [30, 40]]
list2 = list1.copy()

list2[0][0] = 100

print(list1)  # [[100, 20], [30, 40]]
print(list2)  # [[100, 20], [30, 40]]
# 👉 Here, changing list2[0][0] also changed list1[0][0].
# Why? Because both lists share the same inner list.

# Example with Deep Copy
# python
import copy

list1 = [[10, 20], [30, 40]]
list2 = copy.deepcopy(list1)

list2[0][0] = 100

print(list1)  # [[10, 20], [30, 40]]
print(list2)  # [[100, 20], [30, 40]]

# 👉 Now, changing list2 does not affect list1.
# Because deepcopy makes a full independent copy of everything inside.

# 🟢 Easy analogy
# Shallow copy: You photocopy a page, but sticky notes attached to the page are shared. If you change the sticky note on the copy, the original sticky note also changes.

# Deep copy: You photocopy the page and also make new sticky notes. Now both page and notes are independent.
```

```python
import copy
original = [[1, 2], [3, 4]]
copied = copy.deepcopy(original)

copied[0][0] = 100

print(original)
print(copied)
```
