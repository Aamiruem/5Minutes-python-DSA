Why fib[-1] is 1 and fib[-2] is 0 (then 1)
In Python, negative indexes mean "count from the end of the list."

fib[-1] → last element of the list

fib[-2] → second-to-last element of the list

At the very start:
python
fib = [0, 1]
fib[-1] → last element → 1

fib[-2] → second-to-last element → 0

So the first new number is 0 + 1 = 1.

After appending once:
python
fib = [0, 1, 1]
fib[-1] → last element → 1

fib[-2] → second-to-last element → 1

So the next new number is 1 + 1 = 2.

After appending again:
python
fib = [0, 1, 1, 2]
fib[-1] → last element → 2

fib[-2] → second-to-last element → 1

So the next new number is 1 + 2 = 3.

👉 Each time, Python looks at the last two numbers in the list (fib[-1] and fib[-2]) and adds them. That’s why the sequence keeps growing correctly.

```python
# fibonacci
fib = [0, 1]

n = 5
for i in range(n):
    fib.append(fib[-1] + fib[-2])

print(", ".join(str(e) for e in fib))
```

| Step (i) | Last two numbers (``fib[-2]``, ``fib[-1]``) | New number added | List after step |
| --- | --- | --- | --- |
| Start | – | – | [0, 1] |
| i = 0 | (0, 1) | 1 | [0, 1, 1] |
| i = 1 | (1, 1) | 2 | [0, 1, 1, 2] |
| i = 2 | (1, 2) | 3 | [0, 1, 1, 2, 3] |
| i = 3 | (2, 3) | 5 | [0, 1, 1, 2, 3, 5] |
| i = 4 | (3, 5) | 8 | [0, 1, 1, 2, 3, 5, 8] |
