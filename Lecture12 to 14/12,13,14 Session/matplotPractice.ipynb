```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
import matplotlib.pyplot as plt
x = [1, 2,  3, 4, 5, 6]
y = [10, 20, 15, 30, 25, 45]
plt.plot(x, y)

plt.title("Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot([1, 2, 3], [10, 20, 15])
ax.set_title("Simple Graph")
ax.set_xlabel("X Values")
ax.set_ylabel("Y Values ")

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 18, 14, 25, 22]

plt.plot(
    x,
    y,
    color="blue",
    linestyle="--",
    linewidth=3,
    marker="o",
    markersize=8,
    markerfacecolor="yellow",
    markeredgecolor="red",
    alpha=0.8,
    label="Sales"
)

plt.legend()
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales_2025 = [20, 25, 30, 28, 40]
sales_2026 = [25, 30, 27, 35, 45]

plt.plot(months, sales_2025, marker="o", label="2025")
plt.plot(months, sales_2026, marker="s", label="2026")

plt.title("Yearly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.plot([1, 2, 3], [10, 20, 15])

plt.title(
    "Sales Report",
    fontsize=18,
    color="darkblue",
    fontweight="bold",
    loc="center",
    pad=15
)

plt.xlabel("Month", fontsize=15, color="green")
plt.ylabel("Sales", fontsize=15, color="red")

plt.show()
```

```python
plt.plot([1, 2, 3, 4], [10, 30, 20, 40])

plt.grid(
    True,
    axis="both",
    color="gray",
    linestyle="--",
    linewidth=0.7,
    alpha=0.5
)

plt.show()
```
